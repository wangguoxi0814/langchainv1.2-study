# Node总结
- Node是一个运行时对象，一个是一个函数，也可以是合理的Runnable对象，支持同步和异步函数
- Node从源码上来看，会被封装一个可执行对象，并作为`PregelNode.bound`保存
- Node的第一个参数必须是state_schema
- Node可以接收`config: RunnableConfig`,常见用途如下：
    - `config['metadata']`获取元数据，`config["metadata"]["langgraph_step"]`获取当前超步
    - `config[recursion_limit]`获取指定配置数据
    - `config['configurable']['thread_id']`获取会话id
- Node可以接收`runtime: Runtime[{ContextSchema}]`的参数，常见用法：
    - 可以通过`runtime.context`获取运行时上下文
    - 可以通过`runtime.store`获取长期记忆存储器
- Node通常指需要返回发生变化了状态字段，不需要完整的返回
