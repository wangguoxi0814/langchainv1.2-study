# checkpoint、超步、StateSnapshot比较
- 超步分为三阶段，第一阶段计算任务列表，并从检查点恢复状态快照；第二阶段执行任务，产生写入；第三阶段任务完成，保存应用写入，和新检查点（checkpoint_writes和checkpoint）。总的来说，它是一个任务执行的完整周期
- checkpoint是超步结束时保存的状态快照，用于恢复和时间旅行（回溯）
- StateSnapshot也可以理解为是checkpoint+上下文信息，比如tasks就是待执行的任务列表，next。checkpoint是底层存储数据结构，包含checkpoint_id,checkpoint_ns,channel_values等原始字段，而StateSnapshot是用户层视图，通过get_state()\get_state_history可以查看
